```python
#1. univariate visualization
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
#load iris dataset
iris = load_iris()
df=pd.DataFrame(iris.data, columns=iris.feature_names)    
df['species']=iris.target

#histogram
plt.figure(figsize=(6,4))
sns.histplot(df['sepal length (cm)'], bins=10, kde=True)
plt.title("Histogram of Sepal Length")
plt.show()

#boxplot
plt.figure(figsize=(6,4))
sns.boxplot(x=df['sepal length (cm)'])
plt.title("Boxplot of Sepal Length")
plt.show()

#pie chart(species distribution)
plt.figure(figsize=(6,6))
plt.pie(df['species'].value_counts(), labels=iris.target_names, autopct='%1.1f%%')
plt.title("Species Distribution(pie chart)")
plt.show()
```

```python
#2. Bivariate visualization
#scatter plot

plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'], hue=df['species'])
plt.title("Scatterplot:sepal length vs sepal width")
plt.show()
#line chart(trend of sepal length across species)
plt.figure(figsize=(6,4))
plt.plot(df['sepal length (cm)'])
plt.title("Line Chart: Sepal Length")
plt.xlabel("Sample Index")
plt.ylabel("Sepal Length (cm)")
plt.show()

#bar chart(average sepal length by species)
plt.figure(figsize=(6,4))
sns.barplot(data=df, x='species', y='sepal length (cm)')
plt.title("Bar Chart: Sepal Length by Species")
plt.xticks([0, 1, 2], iris.target_names)
plt.show()
```

```python
#3.multivariate visualization
#heatmap(correlation matrix)
plt.figure(figsize=(8,6))   
sns.heatmap(df.iloc[:,:4].corr(), annot=True, cmap='coolwarm')
plt.title("Heatmap of feature Correlations")
plt.show()

#Bubble chart(sepal length vs sepal width with petal length)
plt.figure(figsize=(6,4))
plt.scatter(df['sepal length (cm)'],df['sepal width (cm)'], s=df['petal length (cm)']*20, c=df['species'], alpha=0.6 )
plt.title("Bubble Chart: Sepal Length vs Sepal Width with Petal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.show()

#pairplot(relationships between all features)
sns.pairplot(df.iloc[:,:4])
plt.suptitle("Pairplot of Iris Features", y=1.02)
plt.show()
```

```python
#4. Advanced Techniques
#violin plot
plt.figure(figsize=(6,4))
sns.violinplot(x=df['species'], y=df['sepal length (cm)'])
plt.title("Violin Plot: Sepal Length by Species")   
plt.xticks([0, 1, 2], iris.target_names)
plt.show()
```

```python
# 1. Univariate Visualization for Tips Dataset

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load tips dataset
df = sns.load_dataset("tips")

# Histogram
plt.figure(figsize=(6,4))
sns.histplot(df['total_bill'], bins=10, kde=True)
plt.title("Histogram of Total Bill")
plt.show()

# Boxplot
plt.figure(figsize=(6,4))
sns.boxplot(x=df['total_bill'])
plt.title("Boxplot of Total Bill")
plt.show()

# Pie chart (day distribution)
plt.figure(figsize=(6,6))
plt.pie(df['day'].value_counts(), labels=df['day'].value_counts().index, autopct='%1.1f%%')
plt.title("Day Distribution (Pie Chart)")
plt.show()  
```

```python
# 2. Bivariate Visualization for Tips Dataset

df = sns.load_dataset("tips")

# Scatter plot
plt.figure(figsize=(6,4))

sns.scatterplot(x=df['total_bill'], y=df['tip'], hue=df['sex'])

plt.title("Scatterplot: Total Bill vs Tip")

plt.show()

# Line chart (trend of total bill)
plt.figure(figsize=(6,4))

plt.plot(df['total_bill'])

plt.title("Line Chart: Total Bill")

plt.xlabel("Sample Index")

plt.ylabel("Total Bill")

plt.show()

# Bar chart (average tip by day)
plt.figure(figsize=(6,4))

sns.barplot(data=df, x='day', y='tip')

plt.title("Bar Chart: Tip by Day")

plt.xlabel("Day")

plt.ylabel("Average Tip")

plt.show()
```

```python
# 3. Multivariate Visualization

# Load tips dataset
df = sns.load_dataset("tips")

# Heatmap (correlation matrix)

plt.figure(figsize=(8,6))

sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')

plt.title("Heatmap of Feature Correlations")

plt.show()


# Bubble chart (total bill vs tip with party size)

plt.figure(figsize=(6,4))

plt.scatter(df['total_bill'], df['tip'],
            s=df['size']*30,
            c=df['size'],
            alpha=0.6)

plt.title("Bubble Chart: Total Bill vs Tip with Party Size")

plt.xlabel("Total Bill")

plt.ylabel("Tip")

plt.show()


# Pairplot (relationships between all numerical features)

sns.pairplot(df.select_dtypes(include='number'))

plt.suptitle("Pairplot of Tips Features", y=1.02)

plt.show()
```

```python
#4. Advanced Techniques

#violin plot

plt.figure(figsize=(6,4))

sns.violinplot(x=df['day'], y=df['total_bill'])

plt.title("Violin Plot: Total Bill by Day")

plt.xticks([0, 1, 2, 3], df['day'].unique())

plt.show()
```
