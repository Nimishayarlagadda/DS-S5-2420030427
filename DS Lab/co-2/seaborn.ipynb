```python
#seaborn width and dark grid

import seaborn as sns
import matplotlib.pyplot as plt
data=sns.load_dataset("iris")
sns.lineplot(x="sepal_length",y="sepal_width", data=data)
sns.set_style("darkgrid")
plt.show()
```

```python
# seaborn width

import seaborn as sns
import matplotlib.pyplot as plt
data=sns.load_dataset("iris")
sns.lineplot(x="sepal_length",y="sepal_width", data=data)
sns.set_style("whitegrid")
plt.show()
```

```python
#seaborn width despine

import seaborn as sns
import matplotlib.pyplot as plt
data=sns.load_dataset("iris")
sns.lineplot(x="sepal_length",y="sepal_width", data=data)
sns.despine(left=True, right=False)

plt.show()
```

```python
#kde plot in seaborn
import seaborn as sns
import numpy as np
x=np.random.randn(200)
y=np.random.randn(200)
sns.kdeplot(x)
sns.kdeplot(y)
```

```python
#kde plot in seaborn
import seaborn as sns
import numpy as np
x=np.random.randn(200)
y=np.random.randn(200)
sns.kdeplot(x,shade=True)
plt.grid(False)
sns.kdeplot(y)
```

```python
#2 variables for kde plot in seaborn
import seaborn as sns
import numpy as np
x=np.random.randn(200)
y=np.random.randn(200)
sns.kdeplot(x=x,y=y)
plt.grid(False)
```
