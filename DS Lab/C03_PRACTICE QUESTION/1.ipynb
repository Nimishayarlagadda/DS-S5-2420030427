```python
# bernoulli distrubution
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set the probability of success for the Bernoulli distribution
p = 0.3

# Generate Bernoulli data
data = np.random.binomial(1, p, 1000)

# Plot the histogram
sns.histplot(data, stat="density", discrete=True)

plt.title(f"Bernoulli Distribution (p={p})")
plt.xlabel("Outcome (0 = failure, 1 = success)")
plt.ylabel("Density")
plt.xticks([0, 1])
plt.show()
```

```python
#poisson distribution

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
lam=4
data=np.random.poisson(lam,1000)
sns.histplot(data=data, kde=False, stat="density", bins=range(min(data), max(data)+1))
plt.title(f"Poisson Distribution(lambda={lam})")
plt.xlabel("Number of Events")
plt.ylabel("Density")
plt.show()
```

```python
from scipy.stats import chi2_contingency

# Contingency table
data = [[30, 20],
        [35, 15]]

chi2, p, dof, expected = chi2_contingency(data)

print("Chi-Square Value:", chi2)
print("P-Value:", p)
print("Degrees of Freedom:", dof)
print("Expected Frequencies:")
print(expected)
```

```python
#chi-square test for goodness of fit
from scipy.stats import chisquare

# Observed frequencies
observed = [8, 12, 11, 9, 10, 10]

# Expected frequencies
expected = [10, 10, 10, 10, 10, 10]

chi2, p = chisquare(observed, expected)

print("Chi-Square Value:", chi2)
print("P-Value:", p)
```

```python
import numpy as np
from statsmodels.stats.weightstats import ztest

data = np.random.normal(loc=11.8, scale=0.5, size=100)

population_mean = 12
population_std = 0.5

z_statistic, p_value = ztest(data, value=population_mean)

print(f"Z-Statistic: {z_statistic:.4f}")
print(f"P-Value: {p_value:.4f}")

alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis: The average battery life is different from 12 hours.")
else:
    print("Fail to reject the null hypothesis: The average battery life is not significantly different from 12 hours.")
```

```python
import numpy as np
from scipy import stats

b = np.array([120, 122, 118, 130, 125, 128, 115, 121, 123, 119])
a = np.array([115, 120, 112, 128, 122, 125, 110, 117, 119, 114])

alpha = 0.05

t_stat, p_val = stats.ttest_rel(a, b)

m = np.mean(a - b)
s = np.std(a - b, ddof=1)
n = len(b)

t_manual = m / (s / np.sqrt(n))

decision = "Reject" if p_val <= alpha else "Fail to reject"

concl = "Significant difference." if decision == "Reject" else "No significant difference."

print("T:", t_stat)
print("P:", p_val)
print("T manual:", t_manual)
print(f"Decision: {decision} H0 at α={alpha}")
print("Conclusion:", concl)
```
